```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Создание сетки координат
x = np.linspace(-5, 5, 100)  # 100 точек от -5 до 5 по оси x
y = np.linspace(-5, 5, 100)  # 100 точек от -5 до 5 по оси y

# Создание координатной сетки из x и y
x, y = np.meshgrid(x, y)

# 2. Вычисление значений функции z = sin(sqrt(x^2 + y^2))
z = np.sin(np.sqrt(x**2 + y**2))

# 3. Создание трехмерного графика
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Построение поверхности
surf = ax.plot_surface(x, y, z, cmap='viridis')  # Используем цветовую карту 'viridis'

# 4. Оформление графика
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('График функции z = sin(√(x^2 + y^2))')

# Добавление цветовой панели
fig.colorbar(surf, shrink=0.5, aspect=5)

# 5. Отображение графика
plt.show()
```
