```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Создание данных
x = np.linspace(-10, 10, 100)  # 100 точек от -10 до 10
y = x**2  # Вычисляем y = x^2

# 2. Построение графика
plt.figure(figsize=(8, 6))  # Устанавливаем размер графика (необязательно)
plt.plot(x, y)

# 3. Оформление графика
plt.title("График функции y = x^2", fontsize=16)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.grid(True)  # Включаем сетку

# 4. Отображение графика
plt.show()
```
