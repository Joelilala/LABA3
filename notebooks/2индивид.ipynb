```python
import matplotlib.pyplot as plt
import numpy as np

# Данные
фильмы = ['Фильм A', 'Фильм B', 'Фильм C', 'Фильм D', 'Фильм E']
голоса = [1200, 900, 1500, 800, 1100]

# Цвета для столбцов
цвета = ['red', 'green', 'blue', 'orange', 'purple']

# Построение столбчатой диаграммы
plt.figure(figsize=(10, 6))  # Устанавливаем размер графика

plt.bar(фильмы, голоса, color=цвета) # Используем разные цвета для каждого столбца

# Настройка графика
plt.title('Голосование за лучший фильм', fontsize=16)
plt.xlabel('Фильмы', fontsize=12)
plt.ylabel('Количество голосов', fontsize=12)
plt.grid(axis='y', alpha=0.75) # Сетка только по оси Y

# Отображение значений над столбцами (необязательно)
for i, v in enumerate(голоса):
    plt.text(фильмы[i], v + 30, str(v), ha='center', fontsize=10)

# Показываем график
plt.show()
```
