```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate

# 1. Построение графика подинтегральной функции
x = np.linspace(0, np.pi/2, 100)  # 100 точек на отрезке [0, π/2]
y = np.cos(x)

plt.figure(figsize=(8, 6))
plt.plot(x, y, label='f(x) = cos(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('График f(x) = cos(x) на отрезке [0, π/2]')
plt.grid(True)

# 2. Вычисление площади под кривой (определенного интеграла)
def f(x):
    return np.cos(x)

area, error = integrate.quad(f, 0, np.pi/2)
print("Площадь под кривой:", area)
print("Погрешность:", error)

# 3. Закрашивание области под графиком
plt.fill_between(x, y, color='skyblue', alpha=0.5, label='Площадь под кривой')

plt.legend()
plt.show()
```
