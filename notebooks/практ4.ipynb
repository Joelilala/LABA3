```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Генерация случайных чисел из нормального распределения
mu = 0        # Среднее значение
sigma = 1     # Стандартное отклонение
num_samples = 1000  # Количество случайных чисел
num_bins = 30   # Количество бинов для гистограммы

data = np.random.normal(mu, sigma, num_samples)

# 2. Построение гистограммы
plt.figure(figsize=(8, 6))
plt.hist(data, bins=num_bins, density=True, alpha=0.6, color='skyblue', label='Гистограмма')
#density=True нормализует гистограмму так, чтобы площадь под кривой равнялась 1.

# 3. Добавление вертикальной линии в среднем значении
mean_value = np.mean(data)
plt.axvline(mean_value, color='red', linestyle='dashed', linewidth=2, label=f'Среднее: {mean_value:.2f}')

# 4. Оформление графика
plt.title("Гистограмма случайных чисел из нормального распределения", fontsize=16)
plt.xlabel("Значение", fontsize=12)
plt.ylabel("Плотность", fontsize=12)  # Используем "Плотность" из-за density=True
plt.grid(True)
plt.legend()

# 5. Отображение графика
plt.show()
```
