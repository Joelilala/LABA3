```python
import matplotlib.pyplot as plt

# 1. Данные
оценки = ['Отлично', 'Хорошо', 'Удовлетворительно', 'Неудовлетворительно']
количество_студентов = [20, 35, 30, 15]

# 2. Построение столбчатой диаграммы
plt.figure(figsize=(8, 6))

plt.bar(оценки, количество_студентов, color=['green', 'skyblue', 'orange', 'red'])

# 3. Оформление графика
plt.title("Распределение оценок студентов", fontsize=16)
plt.xlabel("Оценка", fontsize=12)
plt.ylabel("Количество студентов", fontsize=12)
plt.grid(axis='y', alpha=0.75)  # Добавляем сетку по оси Y

# Добавление значений над столбцами (необязательно)
for i, v in enumerate(количество_студентов):
    plt.text(оценки[i], v + 1, str(v), ha='center', fontsize=10)  # Небольшой отступ сверху

# 4. Отображение графика
plt.show()
```
