```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# 1. Создание данных для спиральной поверхности
r = np.linspace(0, 2, 50)  # 50 точек от 0 до 2 для радиуса
theta = np.linspace(0, 4 * np.pi, 50)  # 50 точек от 0 до 4π для угла

# Создание сетки координат из r и theta
r, theta = np.meshgrid(r, theta)

# Вычисление x, y, z координат
x = r * np.cos(theta)
y = r * np.sin(theta)
z = theta

# 2. Визуализация с помощью Matplotlib
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Построение поверхности
surf = ax.plot_surface(x, y, z, cmap='viridis')  # Используем цветовую карту 'viridis'

# 3. Оформление графика
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Спиральная поверхность: z = θ, x = r cos(θ), y = r sin(θ)')

# Добавление цветовой панели (colorbar)
fig.colorbar(surf, shrink=0.5, aspect=5)

# Отображение графика
plt.show()

```
